```python
import numpy as np
import pandas as pd
from linear_regression import LinearRegression
```

```python
df = pd.read_csv('/Users/ianpablogagliardibianchini/Projects/Linear-Regression/housing.csv')

clean_df = pd.get_dummies(df, columns=["ocean_proximity"], drop_first=True)
```

```python
X = clean_df.drop(columns=["median_house_value"]).to_numpy()
y = clean_df["median_house_value"].to_numpy()
```

```python
model = LinearRegression(X, y)
mean_y = model.least_squares_mean()
print(f"Least squares mean:˙{mean_y}")
```
