```python
import pandas as pd
from linear_regression import LinearRegression
```

```python
df = pd.read_csv("housing.csv")
```

```python
df.head()
```

The dataset contains housing information from California.
The goal is to predict median house value

```python
df.info()
df.describe()
```

```python
df = pd.get_dummies(df, columns=["ocean_proximity"], drop_first=True)
df = df.dropna()
X = df.drop("median_house_value", axis=1).to_numpy(dtype=float)
y = df["median_house_value"].to_numpy(dtype=float)
```

All numerical variables are used as features, and median_house_value is used as the target variable. "ocean_proximity" is transformed to dummies so it also can be used as a feature.

```python
model = LinearRegression()
model.fit(X, y)
```

```python
print("Number of samples (n):", model.n)
print("Number of features (d):", model.d)
print("Sample variance:", model.sample_variance(X, y))
print("Standard deviation:", model.standard_deviation(X, y))
print("RMSE:", model.rmse(X, y))
```
